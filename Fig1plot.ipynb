```python
import numpy as np
import matplotlib.pyplot as plt
s=22
plt.figure(figsize=(12, 11))

plt.subplot(2, 2, 1)
plt.text(-0.24, 0.93, "(a)", transform=plt.gca().transAxes, fontsize=s+4)
x = [0,0,0,0]
y = [0,1,0,-1]
plt.scatter(x,y,s=60, color='black')
plt.xlabel('Re$\\beta$', fontsize=s)
plt.ylabel('Im$\\beta$', fontsize=s)
plt.grid()
plt.xlim(-1.1,1.1)
plt.ylim(-1.1,1.1)
plt.xticks([-1,0,1], fontsize=s)
plt.yticks([-1,0,1], fontsize=s)
plt.text(0.53, 0.03, "$\\beta_4$", transform=plt.gca().transAxes, fontsize=s+4)
plt.text(0.53, 0.48, "$\\beta_1,\\beta_3$", transform=plt.gca().transAxes, fontsize=s+4)
plt.text(0.53, 0.93, "$\\beta_2$", transform=plt.gca().transAxes, fontsize=s+4)


plt.subplot(2, 2, 2)
data = np.loadtxt('Eigenvalueq0.txt', dtype=complex)
data2 = np.loadtxt('Eigenvalueq1.txt', dtype=complex)
re, im = np.real(data), np.imag(data)
re2, im2 = np.real(data2), np.imag(data2)
plt.text(-0.2, 0.93, "(b)", transform=plt.gca().transAxes, fontsize=s+4)
plt.scatter(re-8, im, alpha=1, s=60, color='black', marker='x')
#plt.scatter(re2, im2, alpha=1, s=10, color='blue', marker='.')
plt.xlabel('Re$\lambda$', fontsize=s)
plt.ylabel('Im$\lambda$', fontsize=s)
plt.xlim(-9,-7)
plt.xticks([-9,-8.5,-8,-7.5,-7], fontsize=s)
plt.yticks([-4,-2,0,2,4], fontsize=s)
plt.text(0.53, 0.93, "EP3,EP2", transform=plt.gca().transAxes, fontsize=s)
plt.text(0.53, 0.7, "EP4,EP3,EP2", transform=plt.gca().transAxes, fontsize=s)
plt.text(0.53, 0.47, "EP5,EP4,", transform=plt.gca().transAxes, fontsize=s)
plt.text(0.53, 0.39, "EP3,EP2", transform=plt.gca().transAxes, fontsize=s)
plt.grid()

plt.subplot(2, 2, 3)
n1, n2, n3, n4 = np.loadtxt('unnormalize1.txt', dtype=complex), np.loadtxt('unnormalize2.txt', dtype=complex), np.loadtxt('unnormalize3.txt', dtype=complex), np.loadtxt('unnormalize4.txt', dtype=complex)
t = np.arange(0, 5+1e-2, 1e-2)
plt.plot(t, np.exp(4*t*2)*n1, color='black', linewidth=1.3,label='$n_0=1$')
plt.plot(t, np.exp(4*t*2)*n2, color='orange', linewidth=1.3, linestyle='-.', label='$n_0=2$')
plt.plot(t, np.exp(4*t*2)*n3, color='blue', linewidth=1.3, linestyle='--',label='$n_0=3$')
plt.plot(t, np.exp(4*t*2)*n4, color='green', linewidth=1.3, linestyle=':',label='$n_0=4$')
plt.xlabel('t', fontsize=s)
plt.ylabel('exp$(n\gamma t)\\tilde{\langle{N}}\\rangle$', fontsize=s)
plt.xlim(0,4)
plt.ylim(0,100)
plt.xticks([0,1,2,3,4], fontsize=s)
plt.yticks([1,10,50,100], fontsize=s)
plt.text(0.16, 0.48, "$\sim t^3$", transform=plt.gca().transAxes, fontsize=s+4)
plt.text(0.33, 0.36, "$\sim t^2$", transform=plt.gca().transAxes, fontsize=s+4)
plt.text(0.5, 0.15, "$\sim t$", transform=plt.gca().transAxes, fontsize=s+4)
plt.text(0.75, 0.03, "$\sim 1$", transform=plt.gca().transAxes, fontsize=s+4)
plt.legend(fontsize=s, loc='upper right')
plt.tight_layout()
plt.text(-0.27, 0.93, "(c)", transform=plt.gca().transAxes, fontsize=s+4)



plt.subplot(2, 2, 4)
plt.text(-0.15, 0.93, "(d)", transform=plt.gca().transAxes, fontsize=s+4)
t = np.arange(0, 5+1e-2, 1e-2)
n1, n2, n3, n4 = np.loadtxt('Totnumber1.txt', dtype=complex), np.loadtxt('Totnumber2.txt', dtype=complex), np.loadtxt('Totnumber3.txt', dtype=complex), np.loadtxt('Totnumber4.txt', dtype=complex)
plt.plot(t, 1/n1, color='black', linewidth=1.3,label='$n_0=1$')
plt.plot(t, 1/n2, color='orange', linewidth=1.3, linestyle='-.', label='$n_0=2$')
plt.plot(t, 1/n3, color='blue', linewidth=1.3, linestyle='--',label='$n_0=3$')
plt.plot(t, 1/n4, color='green', linewidth=1.3, linestyle=':',label='$n_0=4$')
plt.xlabel('t', fontsize=s)
plt.ylabel('$1/\\tilde{\langle{N}}\\rangle$', fontsize=s)
plt.xlim(0,5)
plt.ylim(0,10)
plt.xticks([0,2,4], fontsize=s)
plt.yticks([1,4,8], fontsize=s)
plt.legend(fontsize=s)
plt.tight_layout()

plt.savefig('Fig1.pdf', dpi=400)
```
